```python
import pandas as pd
import requests
import time
import wget
# inline
```

```python
# https://www.kaggle.com/datasets/dubradave/nfl-draft-history-1990-present (dubradave kaggle dataset)
url = "https://raw.githubusercontent.com/dhoff316/datascience-club-F26/refs/heads/main/NFLDraftHistory.csv"
wget.download(url)
```

```python
# gather hist player data
years = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]
schools = ["Alabama", "Arkansas", "Auburn", "Florida", "Georgia", "Kentucky", "LSU", "Mississippi State", "Missouri", "Oklahoma", "Ole Miss", "South Carolina", "Tennessee", "Texas", "Texas A&M", "Vanderbilt"]
all_data = []

for year in years:
    for school in schools:
        time.sleep(0.5)
        r = requests.get("https://api.collegefootballdata.com/stats/player/season",
                    headers={"Authorization": "Bearer olVvNEK2/yEC2USr1s+zQWta7nLmzM4P1IwXJot9wxU8FIIKob1UPcWq45OEQkXw"},
                    params={
                            "year": year, # loop 2014 - 2025
                            "conference": "SEC",
                            "team": school, # loop schools
                            "seasonType": "both"
                    }
                    )
        data = r.json()
        all_data.extend(data)

df = pd.DataFrame(all_data)
df.to_csv("sec_player_data_2014_2025.csv", index=False)
```

```python
rosters = []

for year in years:
    for school in schools:
        time.sleep(0.5)
        r = requests.get("https://api.collegefootballdata.com/roster",
                    headers={"Authorization": "Bearer olVvNEK2/yEC2USr1s+zQWta7nLmzM4P1IwXJot9wxU8FIIKob1UPcWq45OEQkXw"},
                    params={
                            "team": school,
                            "year": year,
                            "classification": "fbs"
                    }
                    )
        data = r.json()
        rosters.extend(data)

df1 = pd.DataFrame(rosters)
df1.to_csv("sec_roster_data_2014_2025.csv", index=False)
```
